```python
# Imports
import sympy as sp
from sympy import symbols, I, Abs
from sympy.physics.control import TransferFunction
import numpy as np
```

```python
import math
import cmath
```

```python
s, k = sp.symbols('s k')
ganho = k
```

```python
print("Forneça a função de transferência")
numerador_txt = input("Digite o numerador: ") # Digite: s + 1
denominador_txt = input("Digite o denominador: ") # Digite: s**2 + 2*s

numerador_ft = sp.parse_expr(numerador_txt, local_dict={'s': s})
denominador_ft = sp.parse_expr(denominador_txt, local_dict={'s': s})

ft = TransferFunction(numerador_ft, denominador_ft, s)
print(ft)
```

```python
ponto_txt = input("Digite o ponto de teste (a+b*j): ")  # Digite: -3+4*j

ponto_si = sp.parse_expr(ponto_txt, local_dict={'j': sp.I})
print(ponto_si)
```

```python
polos = ft.poles()
zeros = ft.zeros()
print(polos,zeros)
```

```python
diferenca_polos = np.empty((0,2))
diferenca_zeros = np.empty((0,2))

for polo in polos:
  diferenca = ponto_si - polo
  modulo = abs(diferenca)
  fase = math.degrees(cmath.phase(diferenca))
  nova_linha = [[modulo, fase]]
  diferenca_polos = np.append(diferenca_polos, nova_linha, axis=0)

for zero in zeros:
  diferenca = ponto_si - zero
  modulo = abs(diferenca)
  fase = math.degrees(cmath.phase(diferenca))
  nova_linha = [[modulo, fase]]
  diferenca_zeros = np.append(diferenca_zeros, nova_linha, axis=0)

print(diferenca_polos)
print(diferenca_zeros)
```

```python
produto_polos = 1
produto_zeros = 1
theta = 0

for linha in diferenca_polos:
  produto_polos = produto_polos * linha[0]
  theta = theta - linha[1]

for linha in diferenca_zeros:
  produto_zeros = produto_zeros * linha[0]
  theta = theta + linha[1]

M = produto_zeros/produto_polos

# Tratamento para o angulo ser positivo
while theta < 0:
  theta += 360
```

```python
print(f"M = {M}. Theta = {theta}")
```

```python
# Testar se é LGR
while theta > 360:
  theta -= 360

if(theta == 180):
  print("Fase certa, falta testar o K para o módulo")
else:
  print("Não é LGR")
```
